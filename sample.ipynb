# Problem statment

Create a programme that asked for a user integer input, *n* and output the following pattern based on *n*:

$$n=1$$
```
x
```
$$n=2$$
```
x
xx
```
$$n=3$$
```
x
xx
xxx
```

## Sample input
```
1
```
## Sample output
```
X
```
## Sample input
```
2
```
## Sample output
```
X
XX
```
## Sample input
```
3
```
## Sample output
```
X
XX
XXX
```


````python
n = int(input())
for r in range(n):
    print("X"*(r+1))
````
